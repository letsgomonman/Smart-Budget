# Dataset Pengguna

## Import Library & Muat Data Referensi

```python
import pandas as pd
import numpy as np
import uuid
import json

# Set random seed untuk konsistensi
np.random.seed(42)

# Memuat data yang sudah dibuat sebelumnya
df_users = pd.read_csv('dataset_pengguna.csv')
df_transaksi = pd.read_csv('dataset_transaksi.csv')
```

## Sintesis Dataset Anggaran

```python
# Menyiapkan list penampung data
budget_records = []

for _, user in df_users.iterrows():
    user_id = user['user_id']
    pendapatan = user['pendapatan_bulanan']

    # Target tabungan realistis: 10% - 20% dari pendapatan bulanan
    persentase_tabungan = np.random.uniform(0.10, 0.20)
    target_tabungan = int(pendapatan * persentase_tabungan)

    # Batas pengeluaran per kategori (discretionary spending cap)
    # Asumsi: F&B maksimal 15% gaji, Entertainment 10%
    batas_kategori = {
        "Food Delivery": int(pendapatan * 0.15),
        "Coffee Shop": int(pendapatan * 0.05),
        "Movie Ticket": int(pendapatan * 0.05),
        "Gaming": int(pendapatan * 0.05)
    }

    budget_records.append({
        'plan_id': f"PLAN_{str(uuid.uuid4())[:8]}",
        'user_id': user_id,
        'target_tabungan': target_tabungan,
        'batas_pengeluaran_kategori': json.dumps(batas_kategori), # Simpan sebagai JSON String
        'tenggat_waktu': '2023-08-31' # Asumsi target akhir bulan Agustus
    })

# Membangun DataFrame
df_budgets = pd.DataFrame(budget_records)

print("\nPreview Dataset Anggaran:")
display(df_budgets.head())
```

## Sintesis Dataset Anggaran

```python
challenge_records = []
tipe_tantangan_tersedia = ["No Midnight Spending Challenge", "No Coffee Challenge"]

for user_id in df_users['user_id']:
    # Setiap user memiliki peluang 70% untuk mengikuti setidaknya 1 challenge
    ikut_challenge = np.random.choice([True, False], p=[0.7, 0.3])

    if ikut_challenge:
        # Menentukan secara acak apakah user ikut 1 atau 2 challenge sekaligus
        jumlah_challenge = np.random.randint(1, 3)
        tantangan_diikuti = np.random.choice(tipe_tantangan_tersedia, size=jumlah_challenge, replace=False)

        for tantangan in tantangan_diikuti:
            challenge_records.append({
                'challenge_id': f"CHAL_{str(uuid.uuid4())[:8]}",
                'user_id': user_id,
                'tipe_tantangan': tantangan,
                'waktu_mulai': '2023-08-01 00:00:00',
                'waktu_selesai': '2023-08-31 23:59:59',
                'status_tantangan': 'Active' # Status awal selalu Active
            })

# Membangun DataFrame
df_challenges = pd.DataFrame(challenge_records)

print("\nPreview Dataset Tantangan:")
display(df_challenges.head())
```

```python
# Menyimpan ke format CSV
df_budgets.to_csv('dataset_budgets.csv', index=False)
df_challenges.to_csv('dataset_challenge.csv', index=False)
```
